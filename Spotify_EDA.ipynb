### Dataset Overview

This dataset contains Spotify tracks with information about artists, albums, track names, popularity, and audio features. It helps us understand music characteristics and explore patterns in song popularity.

The dataset contains 114,000 rows and 21 columns.

```python

df = df.sample(n=100000, random_state=42)

print("New Dataset Shape:", df.shape)
```

```python

# Check missing values

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python

plt.figure(figsize=(12, 5))

sns.barplot(x=missing_values.index, y=missing_values.values)

plt.title("Missing Values in Spotify Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=90)

plt.show()
```

### Missing Values Analysis

Missing values were checked in all columns of the dataset. A bar chart was created to visualize missing values. Columns with missing values will be handled according to their data types and importance.

```python

# Remove rows with missing values

df = df.dropna()

print("Missing values removed successfully!")
print("Remaining Missing Values:", df.isnull().sum().sum())
print("New Dataset Shape:", df.shape)
```

### Handling Missing Values

The dataset contained 3 missing values in artists, album_name, and track_name. These rows were removed because these columns contain important song information. No missing values remain.

```python
# Check duplicate rows

print("Total Duplicate Rows:", df.duplicated().sum())
```

### Duplicate Rows Analysis

The dataset was checked for duplicate rows. No duplicate records were found. Therefore, no rows were removed during duplicate handling.

### Data Types Analysis

The dataset contains numerical, categorical, and Boolean columns. Numerical columns use integer and float data types, while text columns use object data types. No major data type conversion is required.

```python

df = df.drop(columns=["Unnamed: 0"])

print("Unnecessary column removed!")
print("New Shape:", df.shape)
```

```python

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=df["popularity"])
plt.title("Popularity Outliers")

plt.subplot(1, 2, 2)
sns.boxplot(y=df["duration_ms"])
plt.title("Duration Outliers")

plt.tight_layout()
plt.show()
```

```python

# IQR for Popularity
Q1 = df["popularity"].quantile(0.25)
Q3 = df["popularity"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Popularity Lower Limit:", lower)
print("Popularity Upper Limit:", upper)

print("Popularity Outliers:",
      ((df["popularity"] < lower) | (df["popularity"] > upper)).sum())


# IQR for Duration
Q1 = df["duration_ms"].quantile(0.25)
Q3 = df["duration_ms"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("\nDuration Lower Limit:", lower)
print("Duration Upper Limit:", upper)

print("Duration Outliers:",
      ((df["duration_ms"] < lower) | (df["duration_ms"] > upper)).sum())
```

### Outliers Analysis

The IQR method was applied to popularity and duration_ms. One outlier was found in popularity and 4,925 in duration_ms. These values were kept because unusual song popularity and duration values may be valid in real music data.

### Univariate Analysis

Popularity and danceability distributions were visualized using histograms. A bar chart was used to display the 15 most frequent music genres. These graphs help us understand the distribution of individual variables.

```python

plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="danceability",
    y="popularity",
    alpha=0.3
)

plt.title("Danceability vs Song Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")

plt.show()
```

```python

plt.figure(figsize=(12, 8))

numeric_df = df.select_dtypes(include="number")

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Spotify Features")
plt.show()
```

```python

plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="energy",
    y="loudness",
    alpha=0.3
)

plt.title("Energy vs Loudness")
plt.xlabel("Energy")
plt.ylabel("Loudness")

plt.show()
```

```python

genre_counts = df["track_genre"].value_counts()

print("Most Popular Genre by Track Count:")
print(genre_counts.head(1))

plt.figure(figsize=(10, 5))
sns.barplot(x=genre_counts.head(10).index,
            y=genre_counts.head(10).values)

plt.title("Top 10 Music Genres by Number of Tracks")
plt.xlabel("Music Genre")
plt.ylabel("Number of Tracks")
plt.xticks(rotation=45)
plt.show()
```

The bar chart shows the music genres with the highest number of tracks in the dataset. The genre with the most tracks is identified using value counts.

```python

top_songs = df.nlargest(10, "popularity")

print(top_songs[["track_name", "artists", "popularity"]])

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_songs,
    x="popularity",
    y="track_name"
)

plt.title("Top 10 Most Popular Spotify Songs")
plt.xlabel("Popularity Score")
plt.ylabel("Song Name")

plt.show()
```

The chart displays the top 10 songs with the highest popularity scores. It helps identify which songs are most popular in the dataset.

```python

plt.figure(figsize=(10, 5))

sns.histplot(
    df["popularity"],
    bins=30,
    kde=True
)

plt.title("Distribution of Song Popularity")
plt.xlabel("Popularity Score")
plt.ylabel("Number of Songs")

plt.show()

print("Average Popularity:", df["popularity"].mean())
```

The histogram shows how song popularity scores are distributed. It helps identify the most common popularity ranges and understand the overall popularity pattern.

```python

artist_counts = df["artists"].value_counts().head(10)

print("Top 10 Artists:")
print(artist_counts)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=artist_counts.values,
    y=artist_counts.index
)

plt.title("Top 10 Artists by Number of Tracks")
plt.xlabel("Number of Tracks")
plt.ylabel("Artist Name")

plt.show()
```

The bar chart shows the top 10 artists with the highest number of tracks in the dataset. It identifies artists who have contributed the most tracks.

```python

correlation = df["danceability"].corr(df["popularity"])

print("Danceability and Popularity Correlation:", correlation)

plt.figure(figsize=(10, 5))

sns.regplot(
    data=df,
    x="danceability",
    y="popularity",
    scatter_kws={"alpha": 0.2}
)

plt.title("Danceability vs Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")

plt.show()
```

The scatter plot shows the relationship between danceability and song popularity. Correlation helps measure the strength and direction of this relationship. Correlation does not necessarily mean causation.

```python

correlation = df["energy"].corr(df["loudness"])

print("Energy and Loudness Correlation:", correlation)

plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="energy",
    y="loudness",
    alpha=0.3
)

plt.title("Relationship Between Energy and Loudness")
plt.xlabel("Energy")
plt.ylabel("Loudness (dB)")

plt.show()
```

The scatter plot shows the relationship between song energy and loudness. Correlation helps identify whether higher-energy songs tend to have higher loudness levels.

```python

genre_popularity = (
    df.groupby("track_genre")["popularity"]
    .mean()
    .sort_values(ascending=False)
    .head(15)
)

print("Top 15 Genres by Average Popularity:")
print(genre_popularity)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=genre_popularity.values,
    y=genre_popularity.index
)

plt.title("Average Song Popularity by Music Genre")
plt.xlabel("Average Popularity")
plt.ylabel("Music Genre")

plt.show()
```

The bar chart compares average song popularity across music genres. It helps identify which genres have higher average popularity scores.

```python

explicit_popularity = df.groupby("explicit")["popularity"].mean()

print("Average Popularity:")
print(explicit_popularity)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=explicit_popularity.index.astype(str),
    y=explicit_popularity.values
)

plt.title("Popularity: Explicit vs Non-Explicit Songs")
plt.xlabel("Explicit Content (False / True)")
plt.ylabel("Average Popularity")

plt.show()
```

The bar chart compares the average popularity of explicit and non-explicit songs. It helps identify differences in popularity between the two groups.

```python

# Main Findings

print("1. Most Common Genre:")
print(df["track_genre"].value_counts().idxmax())

print("\n2. Most Popular Song:")
print(df.loc[df["popularity"].idxmax(), "track_name"])

print("\n3. Average Song Popularity:")
print(round(df["popularity"].mean(), 2))

print("\n4. Danceability and Popularity Correlation:")
print(round(df["danceability"].corr(df["popularity"]), 3))

print("\n5. Energy and Loudness Correlation:")
print(round(df["energy"].corr(df["loudness"]), 3))
```
